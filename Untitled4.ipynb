```python
# ============================================================
# AMAZON ML CHALLENGE 2026
# LIGHTWEIGHT INVERTED-INDEX CANDIDATE GENERATION
#
# PURPOSE:
#   1. Download dataset directly in Colab
#   2. Extract without loading whole data into RAM
#   3. Build lightweight token indexes for S2 + S3
#   4. Process ONLY first 10,000 S1 for benchmark
#   5. Generate candidates using:
#        - country partition
#        - rare-token inverted index
#        - name/address evidence
#        - small Top-K
#   6. Measure runtime + RAM + candidate statistics
#
# NO:
#   - Full TF-IDF matrix
#   - Embeddings
#   - FAISS
#   - Pairwise 2M x 10M comparison
#   - Full pandas loading
# ============================================================

!pip -q install duckdb psutil tqdm

import os
import re
import gc
import time
import zipfile
import urllib.request
from collections import defaultdict, Counter

import duckdb
import psutil
import pandas as pd
from tqdm.auto import tqdm


# ============================================================
# 1. CONFIG
# ============================================================

DATASET_URL = "https://cdn.unstop.com/files/6ab10eb3b23ba_student_resource.zip"

BASE_DIR = "/content/amazon_ml"
ZIP_PATH = f"{BASE_DIR}/dataset.zip"
EXTRACT_DIR = f"{BASE_DIR}/dataset"

# IMPORTANT:
# First benchmark only.
TEST_S1_LIMIT = 10_000

# Maximum candidates finally retained per S1.
# We will test this later.
TOP_K = 15

# Token frequency pruning.
# Very common tokens create huge candidate lists.
MAX_TOKEN_DF = 50_000

# Ignore very short tokens.
MIN_TOKEN_LEN = 2

os.makedirs(BASE_DIR, exist_ok=True)


# ============================================================
# 2. DOWNLOAD DATASET
# ============================================================

if not os.path.exists(ZIP_PATH):

    print("Downloading dataset...")

    start = time.time()

    urllib.request.urlretrieve(
        DATASET_URL,
        ZIP_PATH
    )

    print(
        f"Download completed in "
        f"{time.time() - start:.1f} seconds"
    )

else:
    print("Dataset ZIP already exists.")


# ============================================================
# 3. EXTRACT
# ============================================================

if not os.path.exists(EXTRACT_DIR):

    print("Extracting dataset...")

    start = time.time()

    with zipfile.ZipFile(ZIP_PATH, "r") as z:
        z.extractall(EXTRACT_DIR)

    print(
        f"Extraction completed in "
        f"{time.time() - start:.1f} seconds"
    )

else:
    print("Dataset already extracted.")


# ============================================================
# 4. FIND DATASET DIRECTORY AUTOMATICALLY
# ============================================================

print("\nSearching for dataset files...")

required_files = [
    "train_source1.tsv",
    "train_source2.tsv",
    "train_source3.tsv",
    "train_ground_truth.tsv",
    "test_source1.tsv",
    "test_source2.tsv",
    "test_source3.tsv",
]

found = {}

for root, dirs, files in os.walk(EXTRACT_DIR):

    for f in files:

        if f in required_files:
            found[f] = os.path.join(root, f)

print("\nFound files:")

for k, v in found.items():
    print(k, "->", v)


missing = [f for f in required_files if f not in found]

if missing:
    raise FileNotFoundError(
        "Missing files:\n" + "\n".join(missing)
    )


S1_TRAIN = found["train_source1.tsv"]
S2_TRAIN = found["train_source2.tsv"]
S3_TRAIN = found["train_source3.tsv"]
GT_TRAIN = found["train_ground_truth.tsv"]

S1_TEST = found["test_source1.tsv"]
S2_TEST = found["test_source2.tsv"]
S3_TEST = found["test_source3.tsv"]


# ============================================================
# 5. MEMORY MONITOR
# ============================================================

def ram_gb():

    process = psutil.Process(os.getpid())

    return process.memory_info().rss / (1024 ** 3)


print(
    f"\nCurrent Python RAM usage: "
    f"{ram_gb():.2f} GB"
)


# ============================================================
# 6. NORMALIZATION
# ============================================================

def normalize_text(x):

    if x is None:
        return ""

    x = str(x).lower()

    # Replace punctuation with spaces
    x = re.sub(r"[^a-z0-9]+", " ", x)

    # Collapse whitespace
    x = re.sub(r"\s+", " ", x).strip()

    return x


def tokenize(x):

    if not x:
        return []

    return [
        t for t in x.split()
        if len(t) >= MIN_TOKEN_LEN
    ]


# ============================================================
# 7. DUCKDB
# ============================================================

con = duckdb.connect()

print("\nDuckDB ready.")


# ============================================================
# 8. LOAD S2 + S3 IN STREAMING CHUNKS
#
# We do NOT load all 10M rows into pandas.
# ============================================================

print("\nBuilding lightweight inverted index...")


# index:
# country -> token -> list of record references

name_index = defaultdict(list)
address_index = defaultdict(list)

# Track frequency while building.
name_df = Counter()
address_df = Counter()

record_meta = {}

record_count = 0


def process_source(path, source_label):

    global record_count

    print(f"\nProcessing {source_label}")

    # Read using DuckDB in chunks
    query = f"""
        SELECT *
        FROM read_csv(
            '{path}',
            delim='\\t',
            header=true,
            ignore_errors=true,
            null_padding=true
        )
    """

    reader = con.execute(query).fetch_record_batch(
        rows_per_batch=100_000
    )

    for batch in reader:

        df = batch.to_pandas()

        # Detect columns
        cols = {c.lower(): c for c in df.columns}

        id_col = cols.get("entity_id")
        name_col = cols.get("business_name")
        addr_col = cols.get("business_address")
        country_col = cols.get("country")

        if not all([
            id_col,
            name_col,
            addr_col,
            country_col
        ]):
            raise ValueError(
                f"Required columns not found in {source_label}. "
                f"Columns={list(df.columns)}"
            )

        for row in df.itertuples(index=False):

            # tuple -> dictionary-like access
            values = row._asdict()

            entity_id = values[id_col]
            name = normalize_text(values[name_col])
            address = normalize_text(values[addr_col])
            country = normalize_text(values[country_col])

            if not entity_id:
                continue

            # Internal compact ID
            internal_id = record_count

            record_meta[internal_id] = (
                str(entity_id),
                source_label,
                country,
                name,
                address
            )

            # Unique tokens only
            name_tokens = set(tokenize(name))
            address_tokens = set(tokenize(address))

            for token in name_tokens:

                name_df[(country, token)] += 1

            for token in address_tokens:

                address_df[(country, token)] += 1

            record_count += 1

        del df
        gc.collect()

        if record_count % 500_000 == 0:

            print(
                f"Indexed records: {record_count:,} | "
                f"RAM: {ram_gb():.2f} GB"
            )


process_source(S2_TRAIN, "S2")
process_source(S3_TRAIN, "S3")


print(
    f"\nTotal S2+S3 records indexed: "
    f"{record_count:,}"
)

print(
    f"RAM after first pass: "
    f"{ram_gb():.2f} GB"
)


# ============================================================
# 9. SECOND PASS
#
# Build actual inverted index.
#
# We only keep tokens whose frequency is <= MAX_TOKEN_DF.
# This prevents extremely common words from exploding
# candidate counts.
# ============================================================

print("\nBuilding pruned inverted indexes...")

def build_index(path, source_label):

    global_index_count = 0

    query = f"""
        SELECT *
        FROM read_csv(
            '{path}',
            delim='\\t',
            header=true,
            ignore_errors=true,
            null_padding=true
        )
    """

    reader = con.execute(query).fetch_record_batch(
        rows_per_batch=100_000
    )

    # Need mapping from entity_id to internal record.
    # Build lookup once.
    id_lookup = {}

    for iid, meta in record_meta.items():

        entity_id, src, country, name, address = meta

        if src == source_label:

            id_lookup[str(entity_id)] = iid


    for batch in reader:

        df = batch.to_pandas()

        cols = {c.lower(): c for c in df.columns}

        id_col = cols["entity_id"]
        name_col = cols["business_name"]
        addr_col = cols["business_address"]
        country_col = cols["country"]

        for row in df.itertuples(index=False):

            values = row._asdict()

            entity_id = str(values[id_col])

            if entity_id not in id_lookup:
                continue

            iid = id_lookup[entity_id]

            country = normalize_text(
                values[country_col]
            )

            name = normalize_text(
                values[name_col]
            )

            address = normalize_text(
                values[addr_col]
            )

            # Name tokens
            for token in set(tokenize(name)):

                if name_df[(country, token)] <= MAX_TOKEN_DF:

                    name_index[
                        (country, token)
                    ].append(iid)

            # Address tokens
            for token in set(tokenize(address)):

                if address_df[(country, token)] <= MAX_TOKEN_DF:

                    address_index[
                        (country, token)
                    ].append(iid)

            global_index_count += 1

        del df
        gc.collect()

    print(
        f"{source_label} index complete: "
        f"{global_index_count:,} records"
    )


build_index(S2_TRAIN, "S2")
build_index(S3_TRAIN, "S3")


print(
    f"\nName index keys: {len(name_index):,}"
)

print(
    f"Address index keys: {len(address_index):,}"
)

print(
    f"RAM after index construction: "
    f"{ram_gb():.2f} GB"
)


# ============================================================
# 10. CANDIDATE GENERATION
# ============================================================

def generate_candidates(
    country,
    name,
    address,
    top_k=TOP_K
):

    country = normalize_text(country)
    name = normalize_text(name)
    address = normalize_text(address)

    name_tokens = set(tokenize(name))
    address_tokens = set(tokenize(address))

    # Evidence counters
    evidence = Counter()

    # --------------------------------------------------------
    # Name evidence
    # --------------------------------------------------------

    for token in name_tokens:

        key = (country, token)

        for iid in name_index.get(key, []):

            evidence[iid] += 3


    # --------------------------------------------------------
    # Address evidence
    # --------------------------------------------------------

    for token in address_tokens:

        key = (country, token)

        for iid in address_index.get(key, []):

            evidence[iid] += 1


    if not evidence:

        return []


    # --------------------------------------------------------
    # Top candidates
    # --------------------------------------------------------

    return [
        iid
        for iid, score in evidence.most_common(top_k)
    ]


# ============================================================
# 11. PROCESS FIRST 10,000 S1
# ============================================================

print(
    f"\nGenerating candidates for first "
    f"{TEST_S1_LIMIT:,} S1 records..."
)

start_time = time.time()

results = []

candidate_counts = []

query = f"""
    SELECT *
    FROM read_csv(
        '{S1_TRAIN}',
        delim='\\t',
        header=true,
        ignore_errors=true,
        null_padding=true
    )
    LIMIT {TEST_S1_LIMIT}
"""

s1_df = con.execute(query).df()

cols = {c.lower(): c for c in s1_df.columns}

id_col = cols["entity_id"]
name_col = cols["business_name"]
addr_col = cols["business_address"]
country_col = cols["country"]


for row in tqdm(
    s1_df.itertuples(index=False),
    total=len(s1_df)
):

    values = row._asdict()

    s1_id = str(values[id_col])

    country = values[country_col]
    name = values[name_col]
    address = values[addr_col]

    candidates = generate_candidates(
        country,
        name,
        address,
        TOP_K
    )

    candidate_counts.append(len(candidates))

    for iid in candidates:

        entity_id, source, ctry, nm, addr = record_meta[iid]

        results.append({

            "source1_id": s1_id,
            "matched_id": entity_id,
            "source": source,
            "country": ctry,
        })


elapsed = time.time() - start_time


# ============================================================
# 12. RESULTS
# ============================================================

candidate_df = pd.DataFrame(results)

output_path = (
    f"{BASE_DIR}/candidate_pairs_10k.tsv"
)

candidate_df.to_csv(
    output_path,
    sep="\t",
    index=False
)


# ============================================================
# 13. BENCHMARK
# ============================================================

print("\n" + "=" * 60)
print("BENCHMARK RESULT")
print("=" * 60)

print(
    f"S1 processed              : "
    f"{len(s1_df):,}"
)

print(
    f"Candidate pairs generated : "
    f"{len(candidate_df):,}"
)

print(
    f"Average candidates / S1   : "
    f"{sum(candidate_counts) / len(candidate_counts):.2f}"
)

print(
    f"Median candidates / S1    : "
    f"{pd.Series(candidate_counts).median():.2f}"
)

print(
    f"95th percentile           : "
    f"{pd.Series(candidate_counts).quantile(.95):.2f}"
)

print(
    f"Maximum candidates / S1   : "
    f"{max(candidate_counts):,}"
)

print(
    f"Candidate generation time : "
    f"{elapsed:.2f} sec"
)

print(
    f"RAM at end                : "
    f"{ram_gb():.2f} GB"
)

print(
    f"\nOutput:"
)

print(output_path)

print("\nSample:")
display(candidate_df.head(20))
```

```python
# ============================================================
# OPTIMIZED CANDIDATE GENERATION — VERSION 2
# Reuses existing indexes from previous cell
# ============================================================

import time
import gc
import heapq
import pandas as pd
import numpy as np
from tqdm.auto import tqdm


# ============================================================
# CONFIG
# ============================================================

TEST_S1_LIMIT = 10_000

# Test different K later
TOP_K = 15

# Number of rarest tokens to retrieve
MAX_NAME_TOKENS = 2
MAX_ADDRESS_TOKENS = 2

# If a token has too many postings, don't use it
# for retrieval even if it technically exists.
MAX_POSTINGS_TO_SCAN = 20_000


# ============================================================
# PRE-COMPUTE TOKEN FREQUENCY LOOKUPS
# ============================================================

print("Preparing frequency lookups...")

name_freq = name_df
address_freq = address_df


# ============================================================
# FAST TOKEN SELECTION
# ============================================================

def select_rare_tokens(tokens, country, freq_dict, max_tokens):

    valid = []

    for token in tokens:

        df = freq_dict.get(
            (country, token),
            0
        )

        # token must exist in index
        if df <= 0:
            continue

        # skip extremely common tokens
        if df > MAX_POSTINGS_TO_SCAN:
            continue

        valid.append(
            (df, token)
        )

    # rarest first
    valid.sort()

    return [
        token
        for df, token in valid[:max_tokens]
    ]


# ============================================================
# OPTIMIZED CANDIDATE GENERATOR
# ============================================================

def generate_candidates_fast(
    country,
    name,
    address,
    top_k=TOP_K
):

    country = normalize_text(country)
    name = normalize_text(name)
    address = normalize_text(address)

    name_tokens = set(tokenize(name))
    address_tokens = set(tokenize(address))

    # --------------------------------------------------------
    # Select ONLY rarest useful tokens
    # --------------------------------------------------------

    selected_name = select_rare_tokens(
        name_tokens,
        country,
        name_freq,
        MAX_NAME_TOKENS
    )

    selected_address = select_rare_tokens(
        address_tokens,
        country,
        address_freq,
        MAX_ADDRESS_TOKENS
    )

    evidence = {}

    # --------------------------------------------------------
    # NAME RETRIEVAL
    #
    # Name evidence is stronger.
    # --------------------------------------------------------

    for token in selected_name:

        postings = name_index.get(
            (country, token),
            ()
        )

        for iid in postings:

            evidence[iid] = (
                evidence.get(iid, 0) + 3
            )

    # --------------------------------------------------------
    # ADDRESS RETRIEVAL
    # --------------------------------------------------------

    for token in selected_address:

        postings = address_index.get(
            (country, token),
            ()
        )

        for iid in postings:

            evidence[iid] = (
                evidence.get(iid, 0) + 1
            )

    if not evidence:
        return []

    # --------------------------------------------------------
    # Top-K without sorting entire candidate dictionary
    # --------------------------------------------------------

    if len(evidence) <= top_k:

        return [
            x[0]
            for x in sorted(
                evidence.items(),
                key=lambda x: x[1],
                reverse=True
            )
        ]

    return [
        x[0]
        for x in heapq.nlargest(
            top_k,
            evidence.items(),
            key=lambda x: x[1]
        )
    ]


# ============================================================
# LOAD FIRST 10K S1
# ============================================================

print("\nLoading first 10,000 S1...")

query = f"""
    SELECT *
    FROM read_csv(
        '{S1_TRAIN}',
        delim='\\t',
        header=true,
        ignore_errors=true,
        null_padding=true
    )
    LIMIT {TEST_S1_LIMIT}
"""

s1_test_df = con.execute(query).df()

cols = {
    c.lower(): c
    for c in s1_test_df.columns
}

id_col = cols["entity_id"]
name_col = cols["business_name"]
addr_col = cols["business_address"]
country_col = cols["country"]


# ============================================================
# BENCHMARK
# ============================================================

print("\nStarting optimized candidate generation...")

start_time = time.time()

optimized_results = []

candidate_counts = []

zero_candidate_count = 0


for row in tqdm(
    s1_test_df.itertuples(index=False),
    total=len(s1_test_df)
):

    values = row._asdict()

    s1_id = str(values[id_col])

    country = values[country_col]
    name = values[name_col]
    address = values[addr_col]

    candidates = generate_candidates_fast(
        country,
        name,
        address,
        TOP_K
    )

    n = len(candidates)

    candidate_counts.append(n)

    if n == 0:
        zero_candidate_count += 1

    for iid in candidates:

        entity_id, source, ctry, nm, addr = record_meta[iid]

        optimized_results.append({
            "source1_id": s1_id,
            "matched_id": entity_id,
            "source": source,
            "country": ctry
        })


elapsed = time.time() - start_time


# ============================================================
# SAVE
# ============================================================

optimized_df = pd.DataFrame(
    optimized_results
)

optimized_output = (
    "/content/amazon_ml/"
    "candidate_pairs_10k_optimized.tsv"
)

optimized_df.to_csv(
    optimized_output,
    sep="\t",
    index=False
)


# ============================================================
# RESULTS
# ============================================================

stats = pd.Series(candidate_counts)

print("\n")
print("=" * 65)
print("OPTIMIZED BENCHMARK RESULT")
print("=" * 65)

print(
    f"S1 processed              : "
    f"{len(s1_test_df):,}"
)

print(
    f"Candidate pairs generated : "
    f"{len(optimized_df):,}"
)

print(
    f"Average candidates / S1   : "
    f"{stats.mean():.2f}"
)

print(
    f"Median candidates / S1    : "
    f"{stats.median():.2f}"
)

print(
    f"95th percentile           : "
    f"{stats.quantile(.95):.2f}"
)

print(
    f"Maximum candidates / S1   : "
    f"{stats.max():.0f}"
)

print(
    f"Zero-candidate S1         : "
    f"{zero_candidate_count:,}"
)

print(
    f"Candidate generation time : "
    f"{elapsed:.2f} sec"
)

print(
    f"Time                      : "
    f"{elapsed / 60:.2f} min"
)

print(
    f"RAM                       : "
    f"{ram_gb():.2f} GB"
)

print("\nOutput:")
print(optimized_output)

print("\nSample:")
display(optimized_df.head(20))
```

```python
# ============================================================
# STEP 3
# TOP-K BENCHMARK + GROUND TRUTH SCHEMA CHECK
# ============================================================

import time
import pandas as pd
import duckdb
import gc

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

TEST_S1_LIMIT = 10_000
MAX_K = 20

print("=" * 70)
print("STEP 3 — TOP-K BENCHMARK")
print("=" * 70)


# ============================================================
# 1. GENERATE TOP-20 ONLY ONCE
# ============================================================

print("\nGenerating TOP-20 candidates...")

start = time.time()

top20_results = []

for row in tqdm(
    s1_test_df.itertuples(index=False),
    total=len(s1_test_df)
):

    values = row._asdict()

    s1_id = str(values[id_col])

    country = values[country_col]
    name = values[name_col]
    address = values[addr_col]

    candidates = generate_candidates_fast(
        country,
        name,
        address,
        MAX_K
    )

    for rank, iid in enumerate(candidates, start=1):

        entity_id, source, ctry, nm, addr = record_meta[iid]

        top20_results.append({
            "source1_id": s1_id,
            "matched_id": str(entity_id),
            "source": source,
            "country": ctry,
            "rank": rank
        })


top20_df = pd.DataFrame(top20_results)

elapsed = time.time() - start

print(
    f"\nTOP-20 generation time: "
    f"{elapsed:.2f} sec "
    f"({elapsed/60:.2f} min)"
)

print(
    f"Total TOP-20 rows: "
    f"{len(top20_df):,}"
)


# ============================================================
# 2. COMPARE CANDIDATE COUNTS
# ============================================================

print("\n" + "=" * 70)
print("CANDIDATE COUNT BY K")
print("=" * 70)

for k in [5, 10, 15, 20]:

    subset = top20_df[
        top20_df["rank"] <= k
    ]

    avg_candidates = (
        len(subset) /
        TEST_S1_LIMIT
    )

    print(
        f"K={k:<3} "
        f"Candidates={len(subset):>8,} "
        f"Avg/S1={avg_candidates:.2f}"
    )


# ============================================================
# 3. INSPECT GROUND TRUTH SCHEMA
# ============================================================

print("\n" + "=" * 70)
print("GROUND TRUTH SCHEMA")
print("=" * 70)

gt_schema = con.execute(f"""
    DESCRIBE
    SELECT *
    FROM read_csv(
        '{GT_TRAIN}',
        delim='\\t',
        header=true,
        ignore_errors=true,
        null_padding=true
    )
""").df()

display(gt_schema)


# ============================================================
# 4. SHOW FIRST 10 GROUND-TRUTH ROWS
# ============================================================

print("\n" + "=" * 70)
print("GROUND TRUTH SAMPLE")
print("=" * 70)

gt_sample = con.execute(f"""
    SELECT *
    FROM read_csv(
        '{GT_TRAIN}',
        delim='\\t',
        header=true,
        ignore_errors=true,
        null_padding=true
    )
    LIMIT 10
""").df()

display(gt_sample)


# ============================================================
# 5. BASIC DATASET COUNTS
# ============================================================

print("\n" + "=" * 70)
print("GROUND TRUTH BASIC INFO")
print("=" * 70)

gt_count = con.execute(f"""
    SELECT COUNT(*) AS total_rows
    FROM read_csv(
        '{GT_TRAIN}',
        delim='\\t',
        header=true,
        ignore_errors=true,
        null_padding=true
    )
""").fetchone()[0]

print(
    f"Ground truth rows: {gt_count:,}"
)

print(
    "\nNOTE:"
)

print(
    "Ground-truth recall calculation will be done after "
    "checking the exact columns above."
)
```

```python
# ============================================================
# STEP 4 — EXACT CANDIDATE RECALL
# ============================================================
#
# Uses:
#   top20_df       -> generated candidates
#   GT_TRAIN       -> official train ground truth
#   s1_test_df     -> same first 10,000 S1 records
#
# Calculates:
#   K=5,10,15,20
#   - Macro Recall
#   - Micro Recall
#   - True matches
#   - Missed matches
#   - Average candidates / S1
# ============================================================

import pandas as pd
import numpy as np
import time

print("=" * 70)
print("STEP 4 — EXACT CANDIDATE RECALL")
print("=" * 70)


# ============================================================
# 1. GET THE EXACT 10K S1 IDs WE TESTED
# ============================================================

tested_s1_ids = set(
    s1_test_df[id_col]
    .astype(str)
)

print(
    f"\nS1 records being evaluated: "
    f"{len(tested_s1_ids):,}"
)


# ============================================================
# 2. LOAD ONLY GROUND TRUTH FOR THESE 10K S1
#
# Do NOT load all 2.2M GT rows into pandas.
# DuckDB filters it for us.
# ============================================================

print("\nLoading ground truth for these 10K S1...")

con.register(
    "tested_s1_table",
    pd.DataFrame({
        "source1_entity_id": list(tested_s1_ids)
    })
)

gt_10k = con.execute(f"""
    SELECT
        g.source1_entity_id,
        g.matched_entity_ids
    FROM read_csv(
        '{GT_TRAIN}',
        delim='\\t',
        header=true,
        ignore_errors=true,
        null_padding=true
    ) g
    INNER JOIN tested_s1_table t
        ON CAST(g.source1_entity_id AS VARCHAR)
         = t.source1_entity_id
""").df()

print(
    f"Ground-truth S1 rows found: "
    f"{len(gt_10k):,}"
)


# ============================================================
# 3. CONVERT GROUND TRUTH INTO SETS
# ============================================================

def parse_ground_truth(x):

    if pd.isna(x):
        return set()

    x = str(x).strip()

    if not x:
        return set()

    return set(
        item.strip()
        for item in x.split(",")
        if item.strip()
    )


gt_lookup = {}

for row in gt_10k.itertuples(index=False):

    s1_id = str(row.source1_entity_id)

    true_ids = parse_ground_truth(
        row.matched_entity_ids
    )

    gt_lookup[s1_id] = true_ids


# ============================================================
# 4. CREATE PREDICTION LOOKUP
# ============================================================

prediction_lookup = {}

for row in top20_df.itertuples(index=False):

    s1_id = str(row.source1_id)

    prediction_lookup.setdefault(
        s1_id,
        []
    ).append(
        (
            int(row.rank),
            str(row.matched_id)
        )
    )


# ============================================================
# 5. CALCULATE RECALL FOR DIFFERENT K
# ============================================================

print("\n")
print("=" * 70)
print("CANDIDATE RECALL RESULTS")
print("=" * 70)

results = []


for k in [5, 10, 15, 20]:

    total_true = 0
    total_found = 0
    total_missed = 0

    per_s1_recalls = []

    zero_recall_count = 0

    for s1_id, true_ids in gt_lookup.items():

        predictions = prediction_lookup.get(
            s1_id,
            []
        )

        # Only predictions up to K
        predicted_ids = set(
            entity_id
            for rank, entity_id in predictions
            if rank <= k
        )

        found = true_ids.intersection(
            predicted_ids
        )

        missed = true_ids - predicted_ids

        n_true = len(true_ids)
        n_found = len(found)

        total_true += n_true
        total_found += n_found
        total_missed += len(missed)

        # Macro recall:
        # each S1 gets equal importance
        if n_true > 0:

            recall = (
                n_found / n_true
            )

            per_s1_recalls.append(
                recall
            )

            if recall == 0:
                zero_recall_count += 1


    # --------------------------------------------------------
    # MICRO RECALL
    # --------------------------------------------------------

    micro_recall = (
        total_found / total_true
        if total_true > 0
        else 0
    )


    # --------------------------------------------------------
    # MACRO RECALL
    # --------------------------------------------------------

    macro_recall = (
        np.mean(per_s1_recalls)
        if per_s1_recalls
        else 0
    )


    avg_candidates = (
        len(top20_df[
            top20_df["rank"] <= k
        ]) / len(tested_s1_ids)
    )


    results.append({

        "K": k,

        "Avg Candidates/S1":
            avg_candidates,

        "True Matches":
            total_true,

        "Found Matches":
            total_found,

        "Missed Matches":
            total_missed,

        "Micro Recall":
            micro_recall,

        "Macro Recall":
            macro_recall,

        "S1 with 0 Recall":
            zero_recall_count
    })


# ============================================================
# 6. DISPLAY
# ============================================================

recall_df = pd.DataFrame(results)

display(
    recall_df.style.format({
        "Avg Candidates/S1": "{:.2f}",
        "Micro Recall": "{:.4%}",
        "Macro Recall": "{:.4%}"
    })
)


# ============================================================
# 7. SIMPLE READABLE OUTPUT
# ============================================================

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)

for row in results:

    print(
        f"K={row['K']:<2} | "
        f"Avg={row['Avg Candidates/S1']:.2f}/S1 | "
        f"Micro Recall={row['Micro Recall']:.4%} | "
        f"Macro Recall={row['Macro Recall']:.4%} | "
        f"Missed={row['Missed Matches']:,}"
    )


# ============================================================
# 8. SAVE RESULTS
# ============================================================

recall_output = (
    "/content/amazon_ml/"
    "candidate_recall_10k.tsv"
)

recall_df.to_csv(
    recall_output,
    sep="\t",
    index=False
)

print(
    f"\nSaved: {recall_output}"
)
```

```python
# ============================================================
# STEP 5 — ANALYZE WHY TRUE MATCHES ARE MISSED
# ============================================================

import pandas as pd
import numpy as np
from collections import Counter

print("=" * 70)
print("STEP 5 — MISSED MATCH ANALYSIS")
print("=" * 70)


# ============================================================
# 1. BUILD RECORD LOOKUP BY ENTITY ID
# ============================================================

print("\nBuilding S2/S3 entity lookup...")

entity_lookup = {}

for iid, meta in record_meta.items():

    entity_id, source, country, name, address = meta

    entity_lookup[str(entity_id)] = {
        "iid": iid,
        "source": source,
        "country": country,
        "name": name,
        "address": address
    }


# ============================================================
# 2. BUILD TOP-20 PREDICTION SET
# ============================================================

prediction_set = {}

for row in top20_df.itertuples(index=False):

    s1_id = str(row.source1_id)

    prediction_set.setdefault(
        s1_id,
        set()
    ).add(
        str(row.matched_id)
    )


# ============================================================
# 3. BUILD S1 LOOKUP
# ============================================================

s1_lookup = {}

for row in s1_test_df.itertuples(index=False):

    values = row._asdict()

    s1_id = str(values[id_col])

    s1_lookup[s1_id] = {
        "country": normalize_text(
            values[country_col]
        ),
        "name": normalize_text(
            values[name_col]
        ),
        "address": normalize_text(
            values[addr_col]
        )
    }


# ============================================================
# 4. ANALYZE MISSED TRUE MATCHES
# ============================================================

analysis_rows = []

reason_counter = Counter()

total_true = 0
missed_true = 0


for s1_id, true_ids in gt_lookup.items():

    s1 = s1_lookup.get(s1_id)

    if s1 is None:
        continue

    predicted = prediction_set.get(
        s1_id,
        set()
    )

    for true_id in true_ids:

        total_true += 1

        # Already retrieved?
        if true_id in predicted:
            continue

        missed_true += 1

        target = entity_lookup.get(true_id)

        if target is None:

            reason = "target_not_in_index"

            analysis_rows.append({
                "s1_id": s1_id,
                "true_id": true_id,
                "reason": reason,
            })

            reason_counter[reason] += 1

            continue


        # ----------------------------------------------------
        # TOKEN OVERLAP
        # ----------------------------------------------------

        s1_name_tokens = set(
            tokenize(s1["name"])
        )

        s1_addr_tokens = set(
            tokenize(s1["address"])
        )

        target_name_tokens = set(
            tokenize(target["name"])
        )

        target_addr_tokens = set(
            tokenize(target["address"])
        )


        common_name = (
            s1_name_tokens &
            target_name_tokens
        )

        common_address = (
            s1_addr_tokens &
            target_addr_tokens
        )


        # ----------------------------------------------------
        # EXACT NORMALIZED NAME
        # ----------------------------------------------------

        exact_name = (
            s1["name"] != "" and
            s1["name"] == target["name"]
        )


        # ----------------------------------------------------
        # COUNTRY
        # ----------------------------------------------------

        same_country = (
            s1["country"] ==
            target["country"]
        )


        # ----------------------------------------------------
        # CLASSIFY REASON
        # ----------------------------------------------------

        if not same_country:

            reason = "country_mismatch"

        elif exact_name:

            reason = "exact_name_but_not_top20"

        elif common_name:

            reason = "name_token_overlap_but_not_top20"

        elif common_address:

            reason = "address_token_overlap_but_not_top20"

        else:

            reason = "NO_EXACT_TOKEN_OVERLAP"


        reason_counter[reason] += 1


        analysis_rows.append({

            "s1_id": s1_id,

            "true_id": true_id,

            "reason": reason,

            "s1_name": s1["name"],

            "true_name": target["name"],

            "s1_address": s1["address"],

            "true_address": target["address"],

            "common_name_tokens":
                ",".join(
                    sorted(common_name)
                ),

            "common_address_tokens":
                ",".join(
                    sorted(common_address)
                )
        })


# ============================================================
# 5. SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MISSED MATCH SUMMARY")
print("=" * 70)

print(
    f"Total true matches       : "
    f"{total_true:,}"
)

print(
    f"Missed by TOP-20         : "
    f"{missed_true:,}"
)

print(
    f"Miss rate                : "
    f"{missed_true / total_true:.2%}"
)


print("\nReason breakdown:")

for reason, count in reason_counter.most_common():

    pct = (
        count / missed_true
        if missed_true
        else 0
    )

    print(
        f"{reason:<45} "
        f"{count:>8,} "
        f"({pct:.2%})"
    )


# ============================================================
# 6. SAVE DETAILED ANALYSIS
# ============================================================

missed_df = pd.DataFrame(
    analysis_rows
)

missed_output = (
    "/content/amazon_ml/"
    "missed_candidate_analysis_10k.tsv"
)

missed_df.to_csv(
    missed_output,
    sep="\t",
    index=False
)

print(
    f"\nDetailed analysis saved to:"
)

print(missed_output)


# ============================================================
# 7. SHOW IMPORTANT NO-OVERLAP EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("EXAMPLES: NO EXACT TOKEN OVERLAP")
print("=" * 70)

no_overlap = missed_df[
    missed_df["reason"] ==
    "NO_EXACT_TOKEN_OVERLAP"
]

if len(no_overlap) > 0:

    display(
        no_overlap[
            [
                "s1_id",
                "true_id",
                "s1_name",
                "true_name",
                "s1_address",
                "true_address"
            ]
        ].head(20)
    )

else:

    print(
        "No no-overlap examples found."
    )
```

```python
# ============================================================
# STEP 6 — IMPROVED INVERTED-INDEX SCORING
#
# Main change:
#   Previous:
#       name token = +3
#       address token = +1
#
#   New:
#       token rarity
#       multiple token overlaps
#       name/address evidence separately
#       exact normalized name bonus
#
# Existing indexes are reused.
# ============================================================

import math
import heapq
import time
import pandas as pd
import numpy as np
from tqdm.auto import tqdm


# ============================================================
# CONFIG
# ============================================================

TEST_S1_LIMIT = 10_000

TOP_K = 20

MAX_NAME_TOKENS = 5
MAX_ADDRESS_TOKENS = 5

MAX_POSTINGS_TO_SCAN = 50_000


# ============================================================
# FAST RARITY WEIGHT
# ============================================================

def rarity_weight(df):

    if df <= 0:
        return 0.0

    # Rare token = higher weight
    return math.log(
        1.0 +
        1_000_000.0 / df
    )


# ============================================================
# SELECT MORE TOKENS
#
# We now use up to 5 useful tokens instead of 2.
# ============================================================

def select_tokens_v3(
    tokens,
    country,
    freq_dict,
    max_tokens
):

    valid = []

    for token in tokens:

        df = freq_dict.get(
            (country, token),
            0
        )

        if df <= 0:
            continue

        if df > MAX_POSTINGS_TO_SCAN:
            continue

        valid.append(
            (
                df,
                token
            )
        )

    # Rare tokens first
    valid.sort()

    return [
        token
        for df, token in valid[:max_tokens]
    ]


# ============================================================
# IMPROVED CANDIDATE GENERATOR
# ============================================================

def generate_candidates_v3(
    country,
    name,
    address,
    top_k=TOP_K
):

    country = normalize_text(country)
    name = normalize_text(name)
    address = normalize_text(address)

    name_tokens = set(
        tokenize(name)
    )

    address_tokens = set(
        tokenize(address)
    )


    # --------------------------------------------------------
    # Select informative tokens
    # --------------------------------------------------------

    selected_name = select_tokens_v3(
        name_tokens,
        country,
        name_df,
        MAX_NAME_TOKENS
    )

    selected_address = select_tokens_v3(
        address_tokens,
        country,
        address_df,
        MAX_ADDRESS_TOKENS
    )


    # --------------------------------------------------------
    # Candidate evidence
    #
    # candidate:
    #
    # score
    # name_hits
    # address_hits
    # name_rarity
    # address_rarity
    # --------------------------------------------------------

    score = {}
    name_hits = {}
    address_hits = {}


    # --------------------------------------------------------
    # NAME
    # --------------------------------------------------------

    for token in selected_name:

        df = name_df.get(
            (country, token),
            0
        )

        weight = rarity_weight(df)

        postings = name_index.get(
            (country, token),
            ()
        )

        for iid in postings:

            name_hits[iid] = (
                name_hits.get(iid, 0)
                + 1
            )

            score[iid] = (
                score.get(iid, 0.0)
                + 3.0
                + weight
            )


    # --------------------------------------------------------
    # ADDRESS
    # --------------------------------------------------------

    for token in selected_address:

        df = address_df.get(
            (country, token),
            0
        )

        weight = rarity_weight(df)

        postings = address_index.get(
            (country, token),
            ()
        )

        for iid in postings:

            address_hits[iid] = (
                address_hits.get(iid, 0)
                + 1
            )

            score[iid] = (
                score.get(iid, 0.0)
                + 1.5
                + weight
            )


    if not score:

        return []


    # --------------------------------------------------------
    # Additional evidence
    # --------------------------------------------------------

    for iid in list(score.keys()):

        nh = name_hits.get(
            iid,
            0
        )

        ah = address_hits.get(
            iid,
            0
        )


        # Multiple name tokens are strong evidence
        if nh >= 2:

            score[iid] += (
                4.0 *
                (nh - 1)
            )


        # Multiple address tokens
        if ah >= 2:

            score[iid] += (
                2.0 *
                (ah - 1)
            )


        # Exact normalized name
        meta = record_meta[iid]

        target_name = meta[3]

        if (
            name != "" and
            name == target_name
        ):

            score[iid] += 15.0


    # --------------------------------------------------------
    # TOP-K
    # --------------------------------------------------------

    best = heapq.nlargest(
        top_k,
        score.items(),
        key=lambda x: x[1]
    )


    return [
        iid
        for iid, sc in best
    ]


# ============================================================
# RUN ON SAME 10K
# ============================================================

print("=" * 70)
print("STEP 6 — IMPROVED BLOCKER BENCHMARK")
print("=" * 70)

start = time.time()

v3_results = []

for row in tqdm(
    s1_test_df.itertuples(index=False),
    total=len(s1_test_df)
):

    values = row._asdict()

    s1_id = str(values[id_col])

    candidates = generate_candidates_v3(
        values[country_col],
        values[name_col],
        values[addr_col],
        TOP_K
    )

    for rank, iid in enumerate(
        candidates,
        start=1
    ):

        entity_id, source, ctry, nm, addr = (
            record_meta[iid]
        )

        v3_results.append({

            "source1_id":
                s1_id,

            "matched_id":
                str(entity_id),

            "source":
                source,

            "country":
                ctry,

            "rank":
                rank
        })


v3_time = time.time() - start


v3_df = pd.DataFrame(
    v3_results
)


print("\n" + "=" * 70)
print("V3 RESULT")
print("=" * 70)

print(
    f"S1 processed       : {TEST_S1_LIMIT:,}"
)

print(
    f"Candidate rows      : {len(v3_df):,}"
)

print(
    f"Average candidates  : "
    f"{len(v3_df)/TEST_S1_LIMIT:.2f}"
)

print(
    f"Runtime             : "
    f"{v3_time:.2f} sec "
    f"({v3_time/60:.2f} min)"
)

print(
    f"RAM                 : "
    f"{ram_gb():.2f} GB"
)


# ============================================================
# SAVE
# ============================================================

v3_output = (
    "/content/amazon_ml/"
    "candidate_pairs_10k_v3.tsv"
)

v3_df.to_csv(
    v3_output,
    sep="\t",
    index=False
)

print(
    f"\nSaved:"
)

print(v3_output)
```
